```python
point_df = pd.DataFrame(
{
    '1st':[(3,4),(-6,4),(7,-3),(6,9)],
    '2nd':[(2,1),(-6,3),(-9,2),(-3,11)]
})
point_df
def euclidean(row):
    point_a = row['1st']
    point_b = row['2nd']
    return ((point_a[0]-point_b[0])**2 + (point_a[1] - point_b[1])**2)**0.5

point_df.apply(euclidean , axis=1)
point_df['euclidean dis'] = point_df.apply(euclidean , axis=1)
point_df
```
