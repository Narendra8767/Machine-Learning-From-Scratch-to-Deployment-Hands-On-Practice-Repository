```python
#Advance Indexing
#Fancy Indexing
import numpy as np
a1 = np.arange(1,25).reshape(6,4)
print(a1)
print("Row no.1,2,4\n",a1[[0,1,3]])
print("Colum no.1,3,4\n",a1[:,[0,2,3]])
```

```python
#Boolean Indexing
a1 = np.random.randint(1,100,24).reshape(6,4)
print(a1)
#Find all integers greater than 50
print("Greater then 50\n",a1[a1>50])
print("All Even integers\n",a1[a1%2==0])
print("Greater than 50 and even integers\n",a1[(a1>50) & (a1%2==0)])
```

```python
#Broadcasting
#Addition of different shape array
a = np.arange(6).reshape(2,3)
b = np.arange(3).reshape(1,3)
print("First Array\n",a)
print("Second Array\n",b)
print("Addition of Array\n",a+b)
a1 = np.arange(12).reshape(3,4)
b1 = np.arange(4)
print("First Array\n",a1)
print("Second Array\n",b1)
print("Addition of Array\n",a1+b1)
```

```python
#Mathematical Operation in Numpy
#sigmoid
def sigmoid(array):
    return 1/(1 + np.exp(-array))

a = np.arange(9).reshape(3,3)
sigmoid(a)
```

```python
#Mean Squared error in Numpy
actual = np.random.randint(1,50,25)
predicted = np.random.randint(1,50,25)
print("Actual Array\n",actual)
print("Predicted Array\n",predicted)
def mse(actual,predicted):
    return np.mean((actual-predicted)**2)
print("Difference between two Array\n",(actual-predicted))
print("Power of Difference between two Array\n",(actual-predicted)**2)
print("Mean Squared Error\n",mse(actual,predicted))
```

```python
#Working with missing value ->np.nan
a = np.array([1,2,3,4,np.nan,6,7,np.nan,9])
print("Array with missing value\n",a)
print("All missing value\n",a[np.isnan(a)])
print("Delete the missing value from Array\n",a[~np.isnan(a)])
```

```python
#Plotting Graph
import matplotlib.pyplot as plt
x = np.linspace(10,30,15)
y = x
print("X Axis\n",x)
print("Y Axis\n",y)
plt.plot(x,y)
```

```python
x = np.linspace(-20,20,20)
y = x**2
print("X axis\n",x)
print("Y axis\n",y)
plt.plot(x,y)
```

```python
#y = sin(x)
x = np.linspace(-10,10,100)
y = np.sin(x)
print("X axis\n",x)
print("Y axis\n",y)
plt.plot(x,y)
```

```python
#Sort
import numpy as np
a = np.random.randint(1,100,15)
print("1D array\n",a)
print("Sorted Array\n",np.sort(a))
print("Descending Order Array\n",np.sort(a)[::-1])
a1 = np.random.randint(1,100,24).reshape(6,4)
print("2D Array\n",a1)
print("Row wise sorted array\n",np.sort(a1))
print("Colum wise sorted array\n",np.sort(a1,axis=0))
```

```python
#Append the value at end of array
a = np.arange(1,13)
print(a)
print("Add the value at the end of array\n",np.append(a,20))
```

```python
#Concatenate
d1 = np.arange(1,7).reshape(2,3)
d2 = np.arange(11,17).reshape(2,3)
print(d1)
print(d2)
print("concatenate row of two array\n",np.concatenate((d1,d2),axis=0))
print("concatenate colum of two array\n",np.concatenate((d1,d2),axis=1))
```

```python
#unique
e = np.array([1,2,2,2,3,3,4,5,6,6,6,7,7,7,8,9,9])
print(np.unique(e))
```

```python
#percentile
a = np.array([1,54,7,3,6,87,45,76,87,93,65,23,5,78,76,79,43,12,19,53,58,98])
print(np.percentile(a,92))
```
