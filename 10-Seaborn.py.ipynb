```python
tips
```

```python
sns.displot(data=tips, x='total_bill',kind='hist')
```

```python
#bins parameter
sns.displot(data=tips , x='total_bill',kind='hist',bins=20)
```

```python
sns.displot(data=tips,kind='hist',x='sex',hue='sex')
tips['sex'].value_counts()
```

```python
titanic = pd.read_csv('data/Titanic-Train-Data.csv')
titanic
```

```python
sns.displot(data=titanic,kind='hist',x='Age',element='step',hue='Sex')
```

```python
sns.displot(data=tips,kind='hist',x='tip',element='step',col='sex')
```

```python
#kdeplot
#Rather than using discrete bins ,a KDE plot smooth the observations with a Gaussian kernel, producing a continuous density estimate
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True)
```

#Rugplot
#plot marginal distibutions by drawing ticks along the x and y axes
#This function id intended to complate other plts by showing the location of individual observation in an unobtrusive way

```python
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex')
sns.rugplot(data=tips,x='total_bill')
```

```python
#Bivariate histogram
#A bivariate histogram bins the data within rectangles that title the plot
#and then show the counts of observation within each rectangle with the fill color

sns.displot(data=tips,x='total_bill', y='tip',kind='hist')
```

```python
#Bivariate kdeplot
#a bivariate KDE plot smoothes the(x,y) observation with a 2D Gaussian
sns.displot(data=tips,kind='kde',x='total_bill',y='tip')
```
