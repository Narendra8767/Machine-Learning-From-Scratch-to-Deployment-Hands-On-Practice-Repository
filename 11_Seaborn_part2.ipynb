 ## Categorical Plot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
iris = pd.read_csv('data/iris.csv')
tips = pd.read_csv('data/tips.csv')
tips
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip')
```

```python
#Strip plot
sns.catplot(data=tips,x='day',y='total_bill',kind='strip')
```

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='strip',jitter=0.2,hue='sex')
```

```python
#swarmplot
sns.catplot(data=tips,x='day',y='total_bill',kind='swarm',hue='sex')
```

## Categorical Distribution Plot

```python
#Boxplot
sns.catplot(data=tips,x='sex',y='total_bill',kind='box')
```

```python
sns.catplot(data=tips,x='day',y='total_bill',hue='sex',kind='box')
```

```python
#Violinplot=(Boxplot+KDEplot)
```

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='violin',hue='sex')
```

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='violin',hue='sex',split=True)
```

## Categorical Estimate Plot => for central tendency

```python
#Barplot
sns.catplot(data=tips,kind='bar',x='sex',y='total_bill')
```

```python
import numpy as np
sns.catplot(data=tips,kind='bar',x='sex',y='total_bill',estimator=np.min)
```

```python
#Pointplot
sns.catplot(data=tips,kind='point',x='sex',y='total_bill',hue='smoker',errorbar=None)
```

```python
#countplot
sns.countplot(data=tips,x='day',hue='sex')
```

## pointplot

```python
#faceting using catplot
sns.catplot(data=tips,x='sex',y='total_bill',col='smoker',kind='box',row='time')
```

## Regression Plots

```python
#regplot
sns.regplot(data=tips,x='total_bill',y='tip')
```

```python
#lmplot
sns.lmplot(data=tips,x='total_bill',y='tip',hue='sex')
```

```python
#Residual plot
sns.residplot(data=tips,x='total_bill',y='tip')
```

## MultiGrid Plot

```python
#Figure level =>relplot->displot->catplot
#FacetGrid
sns.catplot(data=tips,x='sex',y='total_bill',kind='violin',col='day',row='time')
```

```python
g=sns.FacetGrid(data=tips,col='day',row='time')
g.map(sns.boxplot,'sex','total_bill')
```

```python
#pairplot
sns.pairplot(iris,hue='Species')
```

```python
#pair grid
g = sns.PairGrid(data=iris,hue='Species')
g.map(sns.scatterplot)
```

```python
#map_diag -> map_offdiag
g = sns.PairGrid(data=iris,hue='Species')
g.map_diag(sns.boxplot)
g.map_offdiag(sns.histplot)
```

```python
g = sns.PairGrid(data=iris,hue='Species')
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.histplot)
```

```python
g = sns.PairGrid(data=iris,hue='Species',vars=['SepalWidthCm','PetalLengthCm'])
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.histplot)
```

```python
#Jointplot and JointGrid
sns.jointplot(data=tips,x='total_bill',y='tip',kind='reg')
```
