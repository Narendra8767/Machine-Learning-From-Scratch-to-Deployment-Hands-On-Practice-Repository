### Working with date 

```python
#Change the datatype of date column
date['date'] = pd.to_datetime(date['date'])
```

#### Extract Year

```python
# Extracting year
date['date_year'] = date['date'].dt.year
```

```python
date.head()
```

#### Extract month no

```python
# Extracting month no
date['date_month_no'] = date['date'].dt.month
```

```python
date.head(5)
```

#### Extract month name

```python
date['date_month_name'] = date['date'].dt.month_name()
```

```python
date.head()
```

#### Extract day

```python
# Extracting day
date['date_day'] = date['date'].dt.day
```

```python
date.head()
```

#### Extract day name

```python
date['date_day_name'] = date['date'].dt.day_name()
```

```python
date.head(5)
```

#### Extract day of week

```python
#day of week
date['day_of_week'] = date['date'].dt.dayofweek
```
