Gradient Descent is a fundamental optimization algorithm widely used in machine learning and deep learning to minimize a function, typically a loss function (or cost function), by iteratively moving towards the minimum value of the function. The goal is to find the parameters (weights and biases) that minimize the error between predicted values and actual values.

```python
from sklearn.datasets import make_regression
import numpy as np
import matplotlib.pyplot as plt
```

```python
X,y = make_regression(n_samples=4, n_features=1,n_informative=1,n_targets=1,noise=80,random_state=13)
```

```python
plt.scatter(X,y)
```

```python
#Lets apply OLS
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
reg.fit(X,y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red')
```

```python
#Lets apply Gradient Descent assuming slope is constant m=78.35
#Lets assume the starting value for intercept b = 0
y_pred = ((78.35*X) + 0).reshape(4)
```

```python
#Iteration 1
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred,color='green',label='b=0')
plt.legend()
```

```python
#Iteration 2
m = 78.35
b = 0

loss_slope = -2 * np.sum(y-m*X.ravel() - b)
loss_slope
```

```python
#Lets take the Learning Rate = 0.1
lr = 0.1

step_size = loss_slope*lr
step_size
```

```python
#Calculating the next intercept
b = b - step_size
b
```

```python
y_pred1 = ((78.35 * X) + b).reshape(4)

plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred,color='green',label='b=0')
plt.plot(X,y_pred1,color='black',label='b=20')
plt.legend()
```

```python
#Iteration 3
loss_slope = -2 * np.sum(y-m*X.ravel() - b)
loss_slope
```

```python
step_size = loss_slope*lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred2 = ((78.35 * X) + b).reshape(4)

plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred,color='green',label='b=0')
plt.plot(X,y_pred1,color='black',label='b=20')
plt.plot(X,y_pred2,color='blue',label='b=25')
plt.legend()
```

```python
#Iteration 4
loss_slope = -2 * np.sum(y-m*X.ravel() - b)
loss_slope
```

```python
step_size = loss_slope*lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred3 = ((78.35 * X) + b).reshape(4)

plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred,color='green',label='b=0')
plt.plot(X,y_pred1,color='black',label='b=20')
plt.plot(X,y_pred2,color='blue',label='b=25.11')
plt.plot(X,y_pred2,color='yellow',label='b=25.95')
plt.legend()
```

```python
#create a loop
b = -100
m = 78.35
lr = 0.01

epochs = 100

for i in range (epochs):
    loss_slope = -2 * np.sum(y - m * X.ravel() - b)
    b = b - (lr * loss_slope)
    
    y_pred = m * X + b
    
    plt.plot(X,y_pred)

    
plt.scatter(X,y)    
```
