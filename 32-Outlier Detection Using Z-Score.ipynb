### Calculate Z-Score

```python
df['cgpa_zscore'] = (df['cgpa'] - df['cgpa'].mean()) / df['cgpa'].std()
```

```python
df.head()
```

```python
df[(df['cgpa_zscore'] > 3 ) | (df['cgpa_zscore'] < -3 )]
```

### Capping

```python
upper_limit = df['cgpa'].mean() + 3*df['cgpa'].std()
lower_limit = df['cgpa'].mean() - 3*df['cgpa'].std()
```

```python
lower_limit
```

```python
df['cgpa'] = np.where(
     df['cgpa'] > upper_limit,
     upper_limit,
     np.where(
        df['cgpa']< lower_limit,
        lower_limit,
        df['cgpa']))
```
