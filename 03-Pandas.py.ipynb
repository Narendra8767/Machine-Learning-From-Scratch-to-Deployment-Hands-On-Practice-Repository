## Basic data structures in pandas
1)Series <br>
2)DataFrame

### Series

Series: a one-dimensional labeled array holding data of any type such as integers,string

```python
#From ndarray
s = pd.Series([1, 3, 5, np.nan, 6, 8,9])
s
```

```python
s1 = pd.Series(np.random.random(5),index=["a","b","c","d","e"])
s1
```

```python
#From dict
s2 = {'a':43,'b':76,'c':32,'d':76}
pd.Series(s2)
```

```python
#Name Attribute
s1 = pd.Series(np.random.random(5),name="Random no.")
s1
```

### Creating Dataframe

DataFrame: a two-dimensional data structure that holds data like a two-dimension array or a table with rows and columns.

```python
#using list
student_data = [
    [100,80,10],
    [80,60,7],
    [120,100,14],
    [50,30,3]
]
pd.DataFrame(student_data,columns=['IQ','Marks','Package(LPA)'])
```

## From dict of SeriesTo DataFrane

```python
d = {
    "one":pd.Series([32,4,65,76,34]),
    "Two":pd.Series([65,87,43,88,35]),
    "Three":pd.Series([54,76,46,76,54])
}
df = pd.DataFrame(d)
df
```

```python
df1 = pd.DataFrame(d,columns=["Two","one"])
df1
```

```python
student_info = {
    'Name':["Narendra Tekakel","Balaji Tekale","Priyanka Tekale","Bhagwat Tekale"],
    'Place':["Pune","Parbhani","Pune","Nanded"],
    'Role':["AIML","Mech","MSc","Msc"]
}
pd.DataFrame(student_info)
```
