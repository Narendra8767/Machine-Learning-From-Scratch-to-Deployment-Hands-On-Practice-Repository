```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
X,y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1,noise=20,random_state=13)
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
lr = LinearRegression()
lr.fit(X_train,y_train)
y_pred = lr.predict(X_test)
print(lr.coef_)
print(lr.intercept_)
```

```python
print("r2_score before Ridge Regression",r2_score(y_test,y_pred)*100)
```

```python
from sklearn.linear_model import Ridge
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
rr = Ridge(alpha=5)
rr.fit(X_train,y_train)
print(rr.coef_)
print(rr.intercept_)
```

```python
y_pred1 = rr.predict(X_test)
```

```python
print("r2_score on alpha=5",r2_score(y_test,y_pred1)*100)
```

```python
rr1 = Ridge(alpha=7)
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
rr1.fit(X_train,y_train)
print(rr1.coef_)
print(rr1.intercept_)
```

```python
y_pred2 = rr1.predict(X_test)
```

```python
print("r2_score on alpha=7",r2_score(y_test,y_pred2)*100)
```

```python
rr2 = Ridge(alpha=8)
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
rr2.fit(X_train,y_train)
print(rr2.coef_)
print(rr2.intercept_)
```

```python
y_pred3 = rr2.predict(X_test)
```

```python
print("r2_score on alpha=8",r2_score(y_test,y_pred3)*100)
```

```python
rr3 = Ridge(alpha=10)
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
rr3.fit(X_train,y_train)
print(rr3.coef_)
print(rr3.intercept_)
```

```python
y_pred4 = rr3.predict(X_test)
```

```python
r2_score(y_test,y_pred4)
```

```python
plt.plot(X,y,'b.')
plt.plot(X,lr.predict(X),color='red',label='alpha=0')
plt.plot(X,rr.predict(X),color='green',label='alpha=5')
plt.plot(X,rr1.predict(X),color='orange',label='alpha=7')
plt.plot(X,rr2.predict(X),color='black',label='alpha=8')
plt.legend()
```

## Conclusion:Best fit line is on alpha=7 <br>Most accurate r2_score on alpha=7 is 63.96
