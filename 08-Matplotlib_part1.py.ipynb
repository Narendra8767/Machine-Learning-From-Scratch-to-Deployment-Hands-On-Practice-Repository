```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1)2D Line Plot 
<br>->Bivariate Analysis
<br>->Categorical v/s Numerical And Numerical v/s Numerical
<br>->Use case=Time Series Data

```python
#ploting a simple function
price = [48000,54000,57000,49000,47000,45000]
year = [2015,2016,2017,2018,2019,2020]
plt.plot(year,price)
```

```python
#From a pandas dataframe
batsman = pd.read_csv('data/sharma-kohli.csv')
batsman
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
#ploting multiple plot
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

#GRAPH HEADING
plt.title('V Kohli v/s Rohit Sharma')
#LABLE
plt.xlabel('Season')
plt.ylabel('Runs')
```

```python
#Colors(hex) and line (width and style) and marker(size)
plt.plot(batsman['index'],batsman['V Kohli'],color = 'green',linestyle='dashed',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'],color= 'black',linestyle='dashdot',linewidth=2)

plt.title('V Kohli V/S RG Sharma Career Comparison')
plt.xlabel('Season')
plt.ylabel('Run Scored')
```

```python
#ploting multiple plot
plt.plot(batsman['index'],batsman['V Kohli'],color='red',marker='o')
plt.plot(batsman['index'],batsman['RG Sharma'],color='green',marker='D',markersize=10)

#GRAPH HEADING
plt.title('V Kohli v/s Rohit Sharma')
#LABLE
plt.xlabel('Season')
plt.ylabel('Runs')
```

```python
#marker
plt.plot(batsman['index'],batsman['V Kohli'],marker='o',markersize=10)
plt.plot(batsman['index'],batsman['RG Sharma'],marker='D')

plt.title('V Kohli V/S RG Sharma Career Comparison')
plt.xlabel('Season')
plt.ylabel('Run Scored')
```

```python
#legend -- location
plt.plot(batsman['index'],batsman['V Kohli'],marker='o',label='Virat Kohli')
plt.plot(batsman['index'],batsman['RG Sharma'],marker='D',label='Rohit Sharma')

plt.title('V Kohli V/S RG Sharma Career Comparison')
plt.xlabel('Season')
plt.ylabel('Run Scored')
plt.legend()
#if the there is no label the use plt.legend(['V Kohli',['RG Sharma'])
```

```python
#Outlier...limiting axes
price = [48000,54000,57000,49000,47000,45000,450000]
year = [2015,2016,2017,2018,2019,2020,2021]
plt.plot(year,price)
#limiting the axis
plt.ylim(0,70000)
plt.xlim(2016,2019)
```

```python
#grid
plt.plot(batsman['index'],batsman['V Kohli'],marker='o',label='Virat Kohli')
plt.plot(batsman['index'],batsman['RG Sharma'],marker='D',label='Rohit Sharma')

plt.title('V Kohli V/S RG Sharma Career Comparison')
plt.xlabel('Season')
plt.ylabel('Run Scored')
plt.legend()
plt.grid()
```

## Bar Chart 
<br>->Bivariate Analyasis
<br>->Numerical vs Categorical
<br>->Use Case = Aggregate analysis of groups


```python
children = [10,20,40,10,30]
color = ['red','blue','green','yellow','pink']

plt.bar(color,children)
```

```python
#Horizontal bar chart
plt.barh(color,children)
```

```python
df = pd.read_csv('data/batsman_season_record.csv')
df
```

```python
#Multiple Bar Chart
plt.bar(df['batsman'],df['2015'],width=0.5)
```

```python
plt.bar(np.arange(df.shape[0]) - 0.2,df['2015'],width=0.2,label='2015')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,label='2016')
plt.bar(np.arange(df.shape[0]) + 0.2,df['2017'],width=0.2,label='2017')

plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.legend()
plt.show()
```

```python
#Problem - Label overlapping
plt.xticks(rotation='vertical')
```

```python
#Stacked Bar Chart
plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=df['2016']+ df['2017'],label='2015')
plt.legend()
plt.grid()
```

## Pie Chart
<br>->Univariate/Bivariate Analysis
<br>->Categorical vs Numerical
<br>->Use Case = To find contibution on a standard scale

```python
#simple data
data =[23,45,100,20,49]
subjects=['DBMS','DSA','OOP','AM','FAIML']
plt.pie(data,labels=subjects)
plt.show()
```

```python
df =pd.read_csv('data/gayle-175.csv')
df
```

```python
#autopct='%.2f is use for parcentage in piechart
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%.2f')
plt.show()
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%.2f',explode=[0.1,0,0,0,0,0])
plt.show()
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%.2f',shadow=True)
plt.show()
```
